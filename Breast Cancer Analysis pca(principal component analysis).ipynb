```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
lbs = load_breast_cancer()
```

```python
lbs.keys()
```

```python
#load_breast_cancer, load_iris, etc.) return a Bunch object, which is basically a dictionary-like container.
#It’s not a pandas DataFrame,convert the dataset into a pandas DataFrame first.
df = pd.DataFrame(lbs["data"], columns = lbs["feature_names"])
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
sc.fit(df)
```

```python
dataset =pd.DataFrame(sc.transform(df),columns = df.columns)
```

```python
dataset
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components = 2)
```

```python
pca_result = pca.fit_transform(dataset)
```

```python
pca_df=pd.DataFrame({"pc1":pca_result[:,0],"pc2": pca_result[:,1], "target": lbs["target"]})
```

```python
pca_df
```

```python
sns.scatterplot(x ="pc1",y = "pc2",data =pca_df,hue ="target")
plt.xlabel("first principal component")
plt.ylabel("second principal component") 
plt.show()
```
